# TPBL quick checks
Filter phase and appearance before profiling. Team context repeats per player; do not sum it.

```python
from pathlib import Path
import pandas as pd
root = Path.cwd()
if not (root / 'data').exists(): root = root.parent
p = pd.read_csv(root / 'data/processed/player_game_logs.csv', dtype={'player_id': str, 'game_id': str, 'team_id': str})
t = pd.read_csv(root / 'data/processed/team_game_logs.csv', dtype={'game_id': str, 'team_id': str})
regular = p[(p.competition_phase == 'regular_season') & p.appearance]
assert not p.duplicated(['game_id', 'player_id']).any()
assert set(p.season) == {'2025-26'}
regular.groupby('player_name_original').agg(GP=('game_id','size'), PPG=('points','mean'), RPG=('REB','mean')).sort_values('PPG', ascending=False).head(20)
```
